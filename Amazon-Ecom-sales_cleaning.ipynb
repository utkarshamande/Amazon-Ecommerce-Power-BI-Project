```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv("Downloads/archive (5)/amazon_sales_dataset.csv")
```

```python
df.head()
```

```python
df.info()
```

```python
df.shape
```

```python
#check null values 
df.isnull().sum()
```

```python
# remove duplicates
df.drop_duplicates(inplace=True)
```

```python
# change data type of date col into datetime 
df["order_date "]=pd.to_datetime(df["order_date"])

df[" ship_date "]= pd.to_datetime(df["ship_date"])

df["delivery_date"]= pd.to_datetime(df["delivery_date"])
```

```python
# convert data type 
df['quantity'] = df["quantity"].astype(int)
df["unit_price"] = df["unit_price"].astype(float)
df["discount"] =df["discount"].astype(float)
df["shipping_cost"] = df["shipping_cost"].astype(float)
df["total_sales"]= df["total_sales"].astype(float)


```

```python
df.dtypes
```

```python
#crete profit col
df["profit"]= (df["total_sales"]*0.25)-df["shipping_cost"]
df["profit"]
```

```python
print(df.dtypes)
```

```python
#remove duplicate column order_date, ship_date
df.drop(columns=['order_date','ship_date'], inplace=True)
```

```python
df.dtypes
```

```python
#new col delivery days
df["Delivery days"] = (df["delivery_date"]- df["order_date "]).dt.days

df["Delivery days"]
```

```python
# new col month
df["month"]=df["order_date "].dt.month_name()
df["month"]
```

```python
#new year col
df["Year"]=df["order_date "].dt.year
df["Year"]
```

```python
# Total Revenue 
df["total_sales"].sum()
```

```python
#total categories
df.groupby("category")["total_sales"].sum()
```

```python
#total product 
df.groupby("product_name")["total_sales"].sum().sort_values(ascending=True).head(10)
```

```python
#payment method analysis
df.groupby("payment_method")["total_sales"].sum()
```

```python
#export data 
df.to_csv("cleaned_amazon_sales.csv", index=False)
```
